```python
import numpy as np
import matplotlib.pyplot as plt

fs = 250
time = np.arange(0,5,1/fs)
alpha = np.sin(2*np.pi*10*time)
beta= np.sin(2*np.pi*20*time)
gamma = np.sin(2*np.pi*40*time)
delta = np.sin(2*np.pi*2*time)
theta = np.sin(2*np.pi*6*time)
plt.figure(figsize=(12,10))

plt.subplot(5,1,1)
plt.plot(time,alpha)
plt.title("Alpha 10Hz")
plt.ylabel("Amplitude")

plt.subplot(5,1,2)
plt.plot(time,beta)
plt.title("Beta 20Hz")
plt.ylabel("Amplitude")

plt.subplot(5,1,3)
plt.plot(time,gamma)
plt.title("Gamma 40 Hz")
plt.ylabel("Amplitude")

plt.subplot(5,1,4)
plt.plot(time,delta)
plt.title("Delta 2 Hz")
plt.ylabel("Amplitude")

plt.subplot(5,1,5)
plt.plot(time,theta)
plt.title("Theta 6Hz")
plt.xlabel("time(s)")
plt.ylabel("Amplitude")

plt.tight_layout()
plt.show()
```
